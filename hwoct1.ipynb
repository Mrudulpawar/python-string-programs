### 1.reverse string

```python
name="India"
reverse = ""
for i in name:
    reverse = i + reverse
print("original string:", name)
print("Reverse string:", reverse)
```

### 2.Palindrome String

```python
name=input("Enter a string: ")
if name == name[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

### 3.count vowels from string

```python
name=input("Enter a string: ")
print( name)
count = 0
for i in name:
    if i in "aeiou":
        count+= 1
print("vowels=", count)
```

### 4.count consonents from string


```python
name=input("Enter a string: ")
print(name)
count = 0
for i in name:
    if i not in "aeiou":
        count+= 1
print("consonants=", count)
```

```python
name = input("Enter a string: ")
print(name)
count1 = 0
count2 = 0
for i in name:
    if i in "aeiou":
        count1 += 1
    elif i not in "aeiou":
        count2 += 1
print("vowels=", count1)
print("consonants=", count2)
```

### 8.largest num

```python
no1=int(input("Enter first number: "))
no2=int(input("Enter second number: "))
no3=int(input("Enter third number: "))
if no1>no2 and no1>no3:
    print("Largest number is:", no1)
elif no2>no1 and no2>no3:
    print("Largest number is:", no2)
else:
    print("Largest number is:", no3)
```
